# EDA: PM2.5 e Mortalidade por Poluicao do Ar

Este projeto usa dado real da OMS (WHO Global Health Observatory), nao simulado. PM2.5 medio anual por pais (2010 a 2023) e taxa de mortalidade atribuida a poluicao do ar ambiente (2021, unico ano disponivel para esse indicador). Detalhes em `data/README.md`.

```python
import sys
sys.path.append('..')
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_theme(style='whitegrid', palette='muted')
plt.rcParams['figure.figsize'] = (10, 5)

from src.data.load_data import pipeline_completo

pm25, snapshot = pipeline_completo()
print('serie PM2.5:', pm25.shape, 'paises unicos:', pm25['pais_iso3'].nunique())
print('snapshot 2021:', snapshot.shape)
snapshot.head()
```

## PM2.5 x mortalidade por poluicao, 2021

```python
fig, ax = plt.subplots()
ax.scatter(snapshot['pm25'], snapshot['taxa_mortalidade_poluicao'], alpha=0.6, s=25)
ax.set_xlabel('PM2.5 medio anual (ug/m3)')
ax.set_ylabel('Mortalidade por poluicao do ar (por 100 mil hab.)')
ax.set_title('PM2.5 x mortalidade por poluicao do ar, por pais (2021)')
plt.tight_layout()
plt.savefig('../reports/figures/pm25_x_mortalidade.png', dpi=150)
plt.show()
```

## Conclusoes

Ja da pra ver visualmente uma relacao positiva: paises com PM2.5 mais alto tendem a ter mais mortalidade atribuida a poluicao do ar. Formalizo isso com regressao no proximo notebook.
